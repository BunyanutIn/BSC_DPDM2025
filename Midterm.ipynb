เงื่อนไข:

- ใช้ glob เพื่อค้นหาไฟล์ .csv ทั้งหมดในโฟลเดอร์นั้น

- ใช้ pandas อ่านไฟล์และนำมารวมกันด้วย pd.concat

- มีการจัดการ Error และเช็คว่าโฟลเดอร์หรือไฟล์มีอยู่จริงหรือไม่ (ถ้าไม่พบให้ return DataFrame ว่าง)

- เดือนต้องถูกจัดรูปแบบให้เป็นเลข 2 หลัก (zfill)

- สุดท้ายให้เรียกใช้ฟังก์ชันนี้เพื่อสร้างตัวแปร hii_daily_rainfall07 ถึง hii_daily_rainfall12 สำหรับข้อมูลเดือน กรกฎาคม ถึง ธันวาคม ปี 2025

- รวมทั้ง 6 เดือนในตารางเดียวกัน

```python
import glob
#ใช้สำหรับค้นหาชื่อไฟล์ที่มีลักษณะตามที่เรากำหนด (เช่น ไฟล์ที่ลงท้ายด้วย .csv)
import pandas as pd
#ตัวช่วยหลักในการจัดการข้อมูลในรูปแบบตาราง (DataFrame)
import os
#ใช้จัดการเรื่องที่เกี่ยวกับระบบไฟล์ เช่น การเช็คว่ามีโฟลเดอร์นี้อยู่จริงไหม

def load_daily_rainfall_data(year, month_num):#นิยามฟังก์ชัน รับค่าปี และ ลำดับเดือน
    month_str = str(month_num).zfill(2) #แปลงเลขเดือนเป็นตัวอักษรและเติม 0 ข้างหน้าให้ครบ2 หลัก (เช่น 7 กลายเป็น "07") เพื่อให้ตรงกับชื่อโฟลเดอร์
    directory_path = f'/content/drive/My Drive/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)/{year}-{month_str}/'
    #สร้างที่อยู่ของโฟลเดอร์เป้าหมายโดยใช้ f-string


    all_csv_files = []#เตรียมลิสต์ว่างไว้เก็บรายชื่อไฟล์ที่หาเจอ
    if os.path.isdir(directory_path):#เช็คว่ามีโฟลเดอร์นี้อยู่ใน Drive จริงไหม (กัน Error)
        all_csv_files = glob.glob(os.path.join(directory_path, '*.csv'))#ใช้ glob ไปกวาดชื่อไฟล์ทุกไฟล์ในโฟลเดอร์ที่ลงท้ายด้วย .csv มาเก็บไว้ในลิสต์
    else:
        print(f"Warning: Directory not found for {year}-{month_str}: {directory_path}")#ถ้าไม่เจอโฟลเดอร์ ให้แสดงคำเตือน

    df_list = []#สร้างลิสต์ว่างเพื่อรอเก็บตัวตารางข้อมูลจากแต่ละไฟล์
    if all_csv_files:#ถ้ามีไฟล์ CSV อยู่ในโฟลเดอร์นั้น...
        for file in all_csv_files:#วนลูปอ่านไฟล์ทีละไฟล์
            try:
                df = pd.read_csv(file)
                df_list.append(df)#ถ้าอ่านผ่าน ก็นำตารางนั้นมาใส่ไว้ในลิสต์ df_list
            except Exception as e:
                print(f"Error reading {file}: {e}")
                #ลองอ่านไฟล์ CSV ถ้าไฟล์ไหนเสียหรือมีปัญหาให้ข้ามไปและพิมพ์บอก Error (เพื่อไม่ให้โปรแกรมหยุดทำงานกลางคัน)

        if df_list:#ถ้ามีข้อมูลถูกโหลดมาได้...
            combined_df = pd.concat(df_list, ignore_index=True) #(บรรทัดสำคัญ) เอาตารางทุุกอันในลิสต์มา "วางต่อตูดกัน" (แนวตั้ง) จนกลายเป็นตารางใหญ่แผ่นเดียว
            print(f"Successfully loaded {len(all_csv_files)} files for {year}-{month_str}.")
            return combined_df #ส่งตารางที่รวมเสร็จแล้วกลับออกไป
        else:
            print(f"No dataframes successfully loaded for {year}-{month_str}.")
            return pd.DataFrame() #ถ้าไม่มีข้อมูลเลย ให้ส่งตารางเปล่ากลับไปแทน
    else:
        print(f"No CSV files found for {year}-{month_str} in {directory_path}")
        return pd.DataFrame()

# Load data for each month from July to December 2025
current_year = 2025 #กำหนดปีปัจจุบัน
#เรียกฟังก์ชันเพื่อไปเอาข้อมูลเดือน 7 (กรกฎาคม) มาเก็บในตัวแปรนี้
#(ทำซ้ำแบบเดิมจนถึงเดือน 12)
hii_daily_rainfall07 = load_daily_rainfall_data(current_year, 7)
hii_daily_rainfall08 = load_daily_rainfall_data(current_year, 8)
hii_daily_rainfall09 = load_daily_rainfall_data(current_year, 9)
hii_daily_rainfall10 = load_daily_rainfall_data(current_year, 10)
hii_daily_rainfall11 = load_daily_rainfall_data(current_year, 11)
hii_daily_rainfall12 = load_daily_rainfall_data(current_year, 12)
```

```python
#สร้าง DataFrames all_daily_rainfall_dfs เพื่อรวมข้อมูลทั้ง 6 เดือน
# Create a list of all daily rainfall DataFrames
all_daily_rainfall_dfs = [
    hii_daily_rainfall07,
    hii_daily_rainfall08,
    hii_daily_rainfall09,
    hii_daily_rainfall10,
    hii_daily_rainfall11,
    hii_daily_rainfall12
]

#รวม DataFrame ทั้งหมดเข้าเป็น DataFrame เดียว
# Concatenate all DataFrames into a single one
all_hii_daily_rainfall = pd.concat(all_daily_rainfall_dfs, ignore_index=True)
```

# 3. ระบุข้อมูลสูญหาย(recordที่หายไป)ของแต่ละสถานี แสดงจำนวนและสัดส่วน

```python
import pandas as pd

# 1. Convert 'measure_datetime' to datetime objects and extract the date
# Assuming 'measure_datetime' is the relevant column for daily records.
filtered_hii_daily_rainfall['date'] = pd.to_datetime(filtered_hii_daily_rainfall['measure_datetime']).dt.date

# 2. Define the overall expected date range (July 1, 2025 to December 31, 2025)
start_date = pd.to_datetime('2025-07-01').date()
end_date = pd.to_datetime('2025-12-31').date()
all_expected_dates = pd.date_range(start=start_date, end=end_date, freq='D').date
expected_records_per_station = len(all_expected_dates)

print(f"จำนวนวันที่คาดหวังสำหรับแต่ละสถานี: {expected_records_per_station} วัน (ตั้งแต่ {start_date} ถึง {end_date})")

# 3. Calculate actual unique daily records for each station
station_daily_counts = filtered_hii_daily_rainfall.groupby('tele_station_oldcode')['date'].nunique()

# Convert to DataFrame for easier manipulation
station_data_summary = station_daily_counts.reset_index()
station_data_summary.columns = ['tele_station_oldcode', 'actual_daily_records']

# 4. Compute missing records and proportion
station_data_summary['expected_daily_records'] = expected_records_per_station
station_data_summary['missing_daily_records'] = station_data_summary['expected_daily_records'] - station_data_summary['actual_daily_records']
station_data_summary['missing_proportion'] = station_data_summary['missing_daily_records'] / station_data_summary['expected_daily_records']

# Sort by missing proportion (descending)
missing_data_summary = station_data_summary.sort_values(by='missing_proportion', ascending=False)

print("\nสรุปข้อมูลที่สูญหายของแต่ละสถานี (เรียงตามสัดส่วนที่สูญหายจากมากไปน้อย):")
print(missing_data_summary.head(10))
```

# 5. เติมข้อมูลที่หายไปด้วยค่า Median ตาม Day of Year จาก historical แยกตามสถานี และตรวจสอบว่าไม่มี record หายไป

```python
import pandas as pd

# 1. Merge full_date_template with cleaned_hii_daily_rainfall_processed
#    เพื่อสร้าง DataFrame ที่มีช่วงวันที่ครบถ้วนสำหรับแต่ละสถานี
#    และจะใส่ค่า NaN ในคอลัมน์ 'rainfall_value' สำหรับข้อมูลที่ขาดหายไป
merged_df = pd.merge(
    full_date_template, # DataFrame ต้นฉบับที่มีช่วงวันที่ครบถ้วนสำหรับทุกสถานี
    cleaned_hii_daily_rainfall_processed[['tele_station_oldcode', 'measure_datetime', 'rainfall_value']],
    on=['tele_station_oldcode', 'measure_datetime'], # คอลัมน์ที่ใช้ในการรวมข้อมูล
    how='left' # เก็บทุกแถวจาก full_date_template
)

# 2. Merge ผลลัพธ์ที่ได้กับ historical_medians เพื่อนำค่ามัธยฐานมาใช้ในการเติมข้อมูล
#    เราจะรวมข้อมูลโดยใช้ 'tele_station_oldcode' และ 'day_of_year' เพื่อหาค่ามัธยฐานที่เหมาะสม
merged_df_with_medians = pd.merge(
    merged_df,
    historical_medians[['tele_station_oldcode', 'day_of_year', 'rainfall_value']],
    on=['tele_station_oldcode', 'day_of_year'], # คอลัมน์ที่ใช้ในการรวมข้อมูล
    how='left', # เก็บทุกแถวจาก merged_df
    suffixes=('_current', '_median') # เพิ่ม suffix เพื่อหลีกเลี่ยงชื่อคอลัมน์ที่ซ้ำกัน
)

# 3. เติมค่า 'rainfall_value_current' ที่หายไปด้วยค่าจาก 'rainfall_value_median'
#    โดยการกำหนดค่าโดยตรง เพื่อหลีกเลี่ยง SettingWithCopyWarning และ FutureWarning
imputed_daily_rainfall = merged_df_with_medians.copy() # สร้างสำเนาเพื่อป้องกัน SettingWithCopyWarning
imputed_daily_rainfall['rainfall_value_current'] = imputed_daily_rainfall['rainfall_value_current'].fillna(imputed_daily_rainfall['rainfall_value_median'])

# 4. ลบคอลัมน์ '_median' และเปลี่ยนชื่อคอลัมน์ '_current' เพื่อความชัดเจน
imputed_daily_rainfall.rename(columns={'rainfall_value_current': 'rainfall_value'}, inplace=True)
imputed_daily_rainfall.drop(columns=['rainfall_value_median'], inplace=True)

# 5. ตรวจสอบว่าไม่มีค่าที่หายไปใน 'rainfall_value' หลังจากการเติมข้อมูล
missing_values_after_imputation = imputed_daily_rainfall['rainfall_value'].isnull().sum()
print(f"จำนวนค่าที่หายไปใน 'rainfall_value' หลังจากการเติมข้อมูล: {missing_values_after_imputation}")

# 6. แสดง 5 แถวแรกของ DataFrame ที่ถูกเติมข้อมูลแล้ว
print("\n5 แถวแรกของ DataFrame ที่ถูกเติมข้อมูลแล้ว:")
print(imputed_daily_rainfall.head())

# 7. ตรวจสอบว่าแต่ละสถานีมีจำนวน record ครบถ้วน (ตั้งแต่ 1 ก.ค. ถึง 31 ธ.ค. คือ 184 วัน)
expected_records_per_station = len(pd.date_range(start='2025-07-01', end='2025-12-31'))
actual_records_per_station = imputed_daily_rainfall.groupby('tele_station_oldcode')['measure_datetime'].nunique().reset_index()
actual_records_per_station.columns = ['tele_station_oldcode', 'actual_daily_records']

discrepancies = actual_records_per_station[actual_records_per_station['actual_daily_records'] != expected_records_per_station]

if discrepancies.empty:
    print(f"\nทุกสถานีมีจำนวนบันทึกรายวันครบถ้วนตามที่คาดไว้ ({expected_records_per_station} วัน).")
else:
    print("\nพบความไม่สอดคล้องกันในบันทึกรายวันของบางสถานี:")
    print(discrepancies)

print("\nข้อมูลเกี่ยวกับ imputed_daily_rainfall DataFrame:")
imputed_daily_rainfall.info()
```

# 7.คำนวณผลรวมปริมาณน้ำฝนรายเดือน จัดรูปแบบให้ตรงกับ output_monthly_rainfall.csv

คำนวณผลรวมปริมาณน้ำฝนรายเดือนจาก DataFrame `imputed_daily_rainfall` และจัดรูปแบบผลลัพธ์ให้ตรงกับโครงสร้างของไฟล์ `Y_monthly_rain` (เช่น คอลัมน์ 'Node', 'rain', 'month', 'year')


**Reasoning**:
ขั้นตอนนี้จำเป็นต้องคำนวณผลรวมปริมาณน้ำฝนรายเดือน จึงต้องเริ่มจากการดึงข้อมูล 'year' และ 'month' ออกมาจากคอลัมน์ 'measure_datetime' ใน DataFrame `imputed_daily_rainfall` ก่อน


```python
import pandas as pd

# 1. Extract the 'year' and 'month' from the measure_datetime column
imputed_daily_rainfall['year'] = imputed_daily_rainfall['measure_datetime'].dt.year
imputed_daily_rainfall['month'] = imputed_daily_rainfall['measure_datetime'].dt.month

# 2. Group the imputed_daily_rainfall DataFrame by 'tele_station_oldcode', 'year', and 'month'
# 3. Calculate the sum of 'rainfall_value' for each group to get the monthly rainfall totals.
monthly_rainfall_output = imputed_daily_rainfall.groupby(['tele_station_oldcode', 'year', 'month'])['rainfall_value'].sum().reset_index()

# 4. Rename the columns to 'Node' (from 'tele_station_oldcode') and 'rain' (from 'rainfall_value')
monthly_rainfall_output.rename(columns={'tele_station_oldcode': 'Node', 'rainfall_value': 'rain'}, inplace=True)

# 5. Display the first 5 rows and the information of the resulting DataFrame
print("First 5 rows of the monthly_rainfall_output DataFrame:")
print(monthly_rainfall_output.head())

print("\nInformation about the monthly_rainfall_output DataFrame:")
monthly_rainfall_output.info()
```
